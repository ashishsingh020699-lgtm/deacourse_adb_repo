```sql
CREATE OR REPLACE FUNCTION circuitbox.pii.fn_salary_masking(salary INT) RETURN
CASE WHEN is_account_group_member('admins') THEN salary else "00000" END;
```

```sql
ALTER TABLE employees 
ALTER COLUMN salary SET MASK circuitbox.pii.fn_salary_masking;
```
