```sql
CREATE OR REFRESH MATERIALIZED VIEW circuitbox.gold.customer_order_summary 
AS

SELECT  
c.customer_id
,c.customer_name
,c.date_of_birth
,c.telephone
,c.email
,a.address_line_1
,a.city
,a.state
,a.postcode
,COUNT(DISTINCT(o.order_id)) as total_orders
,SUM(o.item_quantity) as total_items_ordered
,SUM(o.item_price * o.item_quantity) as total_order_amount
FROM
LIVE.circuitbox.lakehouse.silver_customers c
inner join LIVE.circuitbox.silver.addresses a on c.customer_id = a.customer_id
inner join LIVE.circuitbox.silver.orders o on c.customer_name = o.customer_id
WHERE a.__END_AT IS NULL
GROUP BY ALL
```

```sql
DESCRIBE EXTENDED circuitbox.gold.customer_order_summary
```
