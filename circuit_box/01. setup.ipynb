CREATE CATALOG

```sql
CREATE CATALOG IF NOT EXISTS circuitbox
MANAGED LOCATION 'abfss://circuitbox@deacourseadlsa.dfs.core.windows.net/'
COMMENT 'Catalog for circuitbox data lakehouse';
```

CREATE SCHEMAS 

```sql
CREATE SCHEMA IF NOT EXISTS circuitbox.landing
MANAGED LOCATION 'abfss://circuitbox@deacourseadlsa.dfs.core.windows.net/landing'
```

```sql
CREATE SCHEMA IF NOT EXISTS circuitbox.lakehouse
MANAGED LOCATION 'abfss://circuitbox@deacourseadlsa.dfs.core.windows.net/lakehouse'
COMMENT 'this is for processed files'
```

CREATE VOLUME FOR OPERATIONAL DATA

```sql
CREATE EXTERNAL VOLUME IF NOT EXISTS circuitbox.landing.operational_data
  LOCATION 'abfss://circuitbox@deacourseadlsa.dfs.core.windows.net/landing/operational_data'
```
